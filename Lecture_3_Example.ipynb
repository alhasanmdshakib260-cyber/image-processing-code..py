```python
# check width, height and channel
img = skimage.io.imread('/content/dmitry-spravko-yfXHqoFpNH0-unsplash.jpg')
print(f"Width: {img.shape[1]}")
print(f"Height: {img.shape[0]}")
print(f"Channel: {img.shape[2]}")
```

```python
# Show image
plt.figure(figsize=(10, 10))
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.title('Image')
plt.show()
```

```python
# apply gamma correction
gamma = 1.5
g = np.array(255*(img / 255) ** gamma, dtype = 'uint8')

# show gamma corrected image
plt.figure(figsize=(10, 10))
plt.imshow(g, cmap='gray')
plt.axis('off')
plt.title('Gamma Corrected Image')
plt.show()
```

```python
plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.title('Original Image')
plt.subplot(1,2,2)
plt.imshow(g, cmap='gray')
plt.axis('off')
plt.title('Gamma Corrected Image')
plt.show()
```
